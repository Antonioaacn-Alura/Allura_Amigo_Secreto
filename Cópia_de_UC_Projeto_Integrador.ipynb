## Contextualização do problema
Uma organização precisa controlar o acesso a uma área restrita e deseja testar uma solução baseada em reconhecimento facial. O sistema deverá receber uma imagem de entrada, compará-la com uma imagem de referência e indicar se o acesso deve ser autorizado ou negado.

# Objetivo da Aplicação
O objetivo deste notebook é desenvolver uma aplicação educacional baseado em reconhecimento facial. A solução deve carregar imagens, realizar uma comparação facial,interpretar o resultado e gerar uma decisão de acesso Autorizado ou Não-Autorizado.

```python
from deepface import DeepFace
import cv2
import matplotlib.pyplot as plt
import os
```

# 3. Importação das bibliotecas

```python
# Upload das imagens no Google Colab
# Se você não estiver usando Google Colab, coloque as imagens na mesma pasta do notebook.

try:
    from google.colab import files
    print("Ambiente Google Colab detectado.")
    print("Envie agora a imagem de referência e a imagem de teste.")
    uploaded = files.upload()
    print("Arquivos enviados:")
    for nome in uploaded.keys():
        print("-", nome)
except ImportError:
    print("Este ambiente não parece ser o Google Colab.")
    print("Coloque as imagens na mesma pasta deste notebook ou use o caminho completo dos arquivos.")
```

```python
#Listar arquivos:
arquivos = os.listdir()
print("Arquivos disponíveis na pasta atual: ")

for arquivo in arquivos:
  print("-", arquivo)

extensoes_imagem = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
imagens_disponiveis = [a for a in arquivos if a.lower().endswith(extensoes_imagem)]

print("\nPossíveis imagens encontradas: ")
for i, imagem in enumerate(imagens_disponiveis) :
  print(f"{i}: {imagem}")
```

```python
# Seleção das imagens
# Opção 1: se a lista 'imagens_disponiveis' possui pelo menos duas imagens, use os índices.
# Opção 2: se preferir, escreva manualmente os nomes dos arquivos nas variáveis.

if len(imagens_disponiveis) >= 2:
    print("Imagens encontradas automaticamente:")
    for i, imagem in enumerate(imagens_disponiveis):
        print(f"{i}: {imagem}")

    indice_referencia = int(input("Digite o índice da imagem de referência: "))
    indice_teste = int(input("Digite o índice da imagem de teste: "))

    imagem_referencia = imagens_disponiveis[indice_referencia]
    imagem_teste = imagens_disponiveis[indice_teste]
else:
    print("Não foram encontradas duas imagens automaticamente.")
    imagem_referencia = input("Digite o nome ou caminho da imagem de referência: ")
    imagem_teste = input("Digite o nome ou caminho da imagem de teste: ")

print("\nImagem de referência selecionada:", imagem_referencia)
print("Imagem de teste selecionada:", imagem_teste)
```

```python
# Verificar se os arquivos existem
print("Imagem de referência encontrada?", os.path.exists(imagem_referencia))
print("Imagem de teste encontrada?", os.path.exists(imagem_teste))

if not os.path.exists(imagem_referencia):
    raise FileNotFoundError(
        f"A imagem de referência não foi encontrada: {imagem_referencia}. "
        "Verifique se o nome, a extensão e o caminho estão corretos."
    )

if not os.path.exists(imagem_teste):
    raise FileNotFoundError(
        f"A imagem de teste não foi encontrada: {imagem_teste}. "
        "Verifique se o nome, a extensão e o caminho estão corretos."
    )
```

```python
# Leitura das imagens com OpenCV
img_ref = cv2.imread(imagem_referencia)
img_teste = cv2.imread(imagem_teste)

# Validação do carregamento
if img_ref is None:
    raise ValueError(
        "O arquivo da imagem de referência existe, mas não pôde ser lido pelo OpenCV. "
        "Verifique se o arquivo é uma imagem válida."
    )

if img_teste is None:
    raise ValueError(
        "O arquivo da imagem de teste existe, mas não pôde ser lido pelo OpenCV. "
        "Verifique se o arquivo é uma imagem válida."
    )

print("Imagem de referência carregada com sucesso.")
print("Formato da imagem de referência:", img_ref.shape)
print("Imagem de teste carregada com sucesso.")
print("Formato da imagem de teste:", img_teste.shape)
```

```python
# Conversão de BGR para RGB para exibição correta no Matplotlib
img_ref_rgb = cv2.cvtColor(img_ref, cv2.COLOR_BGR2RGB)
img_teste_rgb = cv2.cvtColor(img_teste, cv2.COLOR_BGR2RGB)

# Exibição das imagens
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img_ref_rgb)
plt.title("Imagem de referência")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_teste_rgb)
plt.title("Imagem de teste")
plt.axis("off")

plt.show()
```

```python
# Comparação facial utilizando DeepFace
resultado = DeepFace.verify(
    img1_path=imagem_referencia,
    img2_path=imagem_teste,
    enforce_detection=False
)

print("Resultado completo retornado pela DeepFace:")
resultado
```

```python
# Interpretação do resultado e decisão de acesso
if resultado.get("verified") == True:
    decisao = "Acesso Autorizado"
else:
    decisao = "Acesso Negado"

print("Faces compatíveis:", resultado.get("verified"))
print("Decisão do sistema:", decisao)
```

```python
# Feedback textual para o usuário
if decisao == "Acesso Autorizado":
    print("Acesso Autorizado: a face foi reconhecida como compatível com a referência.")
else:
    print("Acesso Negado: a face não foi reconhecida como compatível com a referência.")
```
